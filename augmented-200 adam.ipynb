```python

import os
import torch
from ultralytics import YOLO
from roboflow import Roboflow

# Check available GPUs
num_gpus = torch.cuda.device_count()
device_ids = list(range(num_gpus))  # Example: [0,1,2,3] if 4 GPUs are available

# Define dataset path (update if needed)
dataset_path = "/home/mansi/Sumit Kumar, Satyendra/Effect of augmentation/Eyes-0.1-2"
data_yaml = os.path.join(dataset_path, "data.yaml")

# Check if dataset exists
if not os.path.exists(data_yaml):
    print("Dataset not found! Downloading...")
    rf = Roboflow(api_key="2kT053S4dqKLHoB6yTpk")
    project = rf.workspace("bitter-gourd-augemented-dataset").project("eyes-0.1")
    version = project.version(2)
    dataset = version.download("yolov11")
    os.rename(dataset.location, dataset_path)
else:
    print("Dataset already exists. Using the existing dataset.")

# Load YOLO model
model = YOLO("yolo11n")  # Adjust model if needed

# Train with optimized settings
model.train(
    data=data_yaml, 
    device=device_ids,  # Use all available GPUs
    epochs=200, 
    batch=32,  # Adjust batch size for better GPU utilization
    workers=8,  # Use multiple data loading workers
    optimizer="Adam"  # Use Adam optimizer
)
```

```python
# Validation
print("Running validation...")
metrics = model.val()
print("Validation Results:", metrics)
```

```python
# Testing
print("Running testing...")
test_results = model.val(split="test")
print("Test Results:", test_results)
```

```python
# Inference
print("Running inference on a sample image...")
image_path = "path_to_your_sample_image.jpg"  # Change this to an actual image path
results = model(image_path)

# Display results
for r in results:
    r.show()  # Show image with detections
    r.save(filename="output.jpg")  # Save output image

print("Inference complete. Results saved as output.jpg")
```
